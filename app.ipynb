# GEN-AI Smart Irrigation System

### AI-powered irrigation recommendation system using Google's Gemini API

This project provides personalized irrigation recommendations for farmers based on:
-  Crop type
-  Soil type
-  Region/Climate
-  Current irrigation method

The system uses the **Google Gemini API** to generate practical, farmer-friendly recommendations that promote efficient water usage.



## Technologies Used
- Python
- Google Gemini API
- ipywidgets
- Jupyter Notebook
- Logging

## Import Required Libraries

```python
import logging
import logging
import os

import google.generativeai as genai
import ipywidgets as widgets

from IPython.display import HTML, Markdown, display
```

## Configure Logging

```python
logging.getLogger("tornado.access").setLevel(logging.CRITICAL)

logger = logging.getLogger("smart_irrigation")
handler = logging.FileHandler("irrigation_errors.log")
handler.setLevel(logging.ERROR)
logger.addHandler(handler)
```

## Configure Gemini API

```python
API_KEY = os.getenv("GEMINI_API_KEY")

genai.configure(api_key=API_KEY)
model = genai.GenerativeModel("models/gemini-1.5-flash")
```

## Create User Interface

```python
display(HTML("""
<div style="
    background: linear-gradient(90deg, #b5f7b5, #e0fce0);
    padding: 25px;
    border-radius: 15px;
    margin-bottom: 20px;
    font-family: 'Segoe UI', sans-serif;">
    <h2 style="color:#2e7d32; font-size:28px; text-align: center;">
        🌱 Smart Irrigation Advisor
    </h2>
    <p style="color:#33691e; font-size:16px; text-align: center;">
        Personalized water-saving tips for farmers — powered by Generative AI 💧
    </p>
</div>
"""))

style = {'description_width': 'initial'}
layout = widgets.Layout(width='400px', height='40px', font_size='18px')
   
```

## Create Input Widgets

```python
crop_input = widgets.Text(
    description="Crop Type:",
    placeholder="e.g., Sugarcane, Rice, Tomato",
    layout=layout,
    style=style
)

soil_input = widgets.Dropdown(
    description="Soil Type:",
    options=['Clay', 'Sandy', 'Loamy', 'Silty', 'Peaty', 'Chalky'],
    layout=layout,
    style=style
)

region_input = widgets.Dropdown(
    description="Region/Climate:",
    options=[
        'Maharashtra', 'Karnataka', 'Punjab', 'Tamil Nadu',
        'Odisha', 'Kerala','Bihar', 'Rajasthan', 'Gujarat', 'Uttar Pradesh', 'West Bengal'
    ],
    layout=layout,
    style=style
)

method_input = widgets.Dropdown(
    description="Current Irrigation Method:",
    options=['Flood', 'Drip', 'Furrow', 'Sprinkler', 'Canal', 'Manual (Bucket)', 'Rain-fed'],
    layout=layout,
    style=style
)

submit_button = widgets.Button(
    description="🌾 Get Advice",
    button_style='success',
    layout=widgets.Layout(width='400px', height='45px')
)
output = widgets.Output()
```

## Generate AI Recommendations

```python
def generate_irrigation_advice(b):
    output.clear_output()

    with output:
        display(Markdown("⏳ **Generating advice... please wait...**"))

    prompt = f"""
    You are an expert agricultural advisor chatbot helping farmers conserve water.

    A farmer grows {crop_input.value} in the {region_input.value} region.
    Their soil type is {soil_input.value}, and they currently use {method_input.value} irrigation.

    Provide the following:
    - An optimal watering schedule
    - A better irrigation technique suited to the crop and soil
    - Advice on how to transition to drip or mulch-based systems
    - Estimated water savings compared to their current method

    Keep the tone simple, clear, and farmer-friendly.
    """

    try:
        response = model.generate_content(prompt)
        output.clear_output()
        with output:
            display(HTML("<hr style='border: 1px solid #ccc;'>"))
            display(Markdown(f"<div style='font-size:18px;'>🌾 **Irrigation Advice**:\n\n{response.text.strip()}</div>"))
    except Exception as e:
        output.clear_output()
        with output:
            display(Markdown("❌ **An error occurred. Please try again later.**"))
        logger.error("Error while generating content", exc_info=True)
```

## Launch Application

```python
submit_button.on_click(generate_irrigation_advice)
form = widgets.VBox([
    crop_input,
    soil_input,
    region_input,
    method_input,
    submit_button,
    output
])

display(form)
```
